```python
import numpy as np

# Unidades de medida

um = 1e-6
nm = 1e-9
```

```python
nc = 1.5
nr = 3.5
lambda_0 = 1550 *nm
k_0 = 2*np.pi/lambda_0
W_m = 7*um
```

```python
sigma = 0 # TE mode

We = W_m + (lambda_0/np.pi)*((nc/nr)**(2*sigma))*(nr**2-nc**2)**(-0.5)

beta = np.zeros(2)

for i in range(2):

    beta[i] = nr*k_0 - ((i+1)**2 * np.pi * lambda_0)/(We**2 *4*nr)

Lpi = np.pi / (beta[0] - beta[1])
L = (3*Lpi)/2

print("Comprimento central = %.4f microns"  %(L*1e6))
```
